## Logistic regression code using sigmoid fcutnion


```python
import numpy as np
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
import seaborn as sns
import matplotlib as mt


import pandas as pd
```

```python
def sigmoid(z):
    return 1/(1+np.exp(-z))
```

```python
X, Y = make_classification(n_samples=100, n_features=3, n_informative=1, n_redundant=1, n_classes=2, n_clusters_per_class=1, random_state=42, hypercube=False)
```

```python
def perceptron(X, Y, lr=0.1):
    if hasattr(X):
        X = X.values
    if hasattr(Y):
        Y = Y.values
    X = np.insert(X, 0 , 1, axis=1)
    w = np.ones(X.shape[1])
    for i in range(1000):
        j = np.random.randint(X.shape[0])
        y = sigmoid(np.dot(X[j], w))
        w = lr*(Y[j] - y)*X[j]
    return w[0], w[1:]
```
