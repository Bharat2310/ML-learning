# KMeans clustering from scratch 


```python
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
```

```python
centroids = [(-10,-5),(1,5)]
cluster_std = [1,1]
X, Y = make_blobs(n_samples=200, cluster_std=cluster_std, centers= centroids, n_features=2, random_state=42 )
```

```python
plt.scatter(X[:, 0], X[:, 1])
plt.show()
```
